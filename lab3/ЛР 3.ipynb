```python
import sympy as sp
```

```python
# 1-3. Инициализация символьных переменных и функций
x, y = sp.symbols('x y', real=True)
lam = sp.symbols('lambda', real=True)

var_list = [x, y]

# Целевая функция и ограничение для варианта №9
target_func = x * y
eq_constraint = x**2 + y**2 - 2

eps = 1e-8 # Погрешность для проверки невязки

# 4-6. Построение функции Лагранжа
lagrangian = target_func + lam * eq_constraint

# Формирование системы уравнений (градиент Лагранжиана по x, y и само ограничение)
system_of_eqs = [sp.diff(lagrangian, v) for v in var_list] + [eq_constraint]

# Вывод исходных данных
print("Целевая функция:", target_func)
print("Ограничения:", eq_constraint, "= 0")
print("Функция Лагранжа:", lagrangian)
print("Уравнения стационарности:")
for idx, equation in enumerate(system_of_eqs, 1):
    print(f"  {idx}) {equation} = 0")
print()

# 7-12. Поиск решений и анализ кандидатов
all_symbols = var_list + [lam]
# Символьное решение системы
raw_solutions = sp.solve(system_of_eqs, all_symbols, dict=True)

extremum_values = [] # Список для хранения значений функции в допустимых точках

for i, current_sol in enumerate(raw_solutions, 1):
    # Исключаем комплексные корни, если они появляются
    if any(val.has(sp.I) for val in current_sol.values()):
        continue

    # Координаты текущей точки для подстановки
    point_coords = {x: current_sol[x], y: current_sol[y]}

    # Вычисление невязки и проверка условия допустимости
    current_residual = sp.simplify(eq_constraint.subs(point_coords))
    is_point_valid = abs(current_residual.evalf()) <= eps

    # Расчет значения целевой функции
    f_value = sp.simplify(target_func.subs(point_coords))

    # Подготовка данных для вывода
    pt_tuple = tuple(current_sol[v] for v in var_list)
    lambda_val = current_sol[lam]

    print(f"Кандидат {i}:")
    print(f"  точка: {pt_tuple}")
    print(f"  множители: {lambda_val}")
    print(f"  значение функции: {f_value}")
    print(f"  невязки ограничений: {current_residual}")
    print(f"  допустимость: {'да' if is_point_valid else 'нет'}\n")

    if is_point_valid:
        extremum_values.append(f_value)

print()
print("Итог:")

# 13. Классификация найденных экстремумов
if extremum_values:
    print(f"  глобальный минимум: {min(extremum_values)}")
    print(f"  глобальный максимум: {max(extremum_values)}")
    print("  основание классификации: множество допустимых значений (окружность) является замкнутым и ограниченным компактом. Целевая функция непрерывна. Благодаря регулярности ограничений, достаточно сравнить значения функции во всех найденных стационарных точках, чтобы выявить глобальные экстремумы.")
else:
    print("  Допустимые точки-кандидаты не найдены.")
```
